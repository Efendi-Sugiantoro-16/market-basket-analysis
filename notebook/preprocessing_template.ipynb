## Data Cleaning
Dua pembersihan krusial:
1. **Menghapus Kolom Noise** (`Unnamed: 6`).
2. **Memperbaiki Tipe Data** — Nilai berat kolom `Kg` (koma → titik → float).

```python
# 1. Bersihkan Kolom Sampah (Noise)
if 'Unnamed: 6' in df_cleaned.columns:
    df_cleaned = df_cleaned.drop(columns=['Unnamed: 6'])
    print('Kolom Unnamed: 6 berhasil dihapus.')

# 2. Perbaiki Format tipe data Kg (Koma -> Titik -> Float)
if df_cleaned['Kg'].dtype == 'object':
    df_cleaned['Kg'] = df_cleaned['Kg'].str.replace(',', '.').astype(float)
    print('Tipe data Kg berhasil dikonversi ke float.')

print('\nTabel Data Transaksi yang Sudah Bersih:')
display(df_cleaned.head())
```

💡 **Pembahasan:**
- Tabel di atas kini sudah bersih dari kolom hantu (`Unnamed: 6`).
- Tipe data pada kolom `Kg` telah distandarisasi menggunakan format titik `.` untuk desimal.
- Data siap untuk tahap eksplorasi dan analisis lebih lanjut.
