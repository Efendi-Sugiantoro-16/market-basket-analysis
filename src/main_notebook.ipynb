```python
from data_loader import load_transaction_raw, load_preprocessing, load_pricelist
from model import create_binary_matrix, run_apriori, generate_rules
from utils import (
    plot_revenue_combo, plot_revenue_item,
    plot_heatmap, plot_top_rules, plot_network,
    print_summary, format_rupiah
)
print('Semua modul berhasil diimpor!')
```

## 1. Load Data

```python
df_raw = load_transaction_raw()
df_pre, transactions = load_preprocessing()
df_price = load_pricelist()

print(f'Transaksi mentah  : {df_raw.shape}')
print(f'Preprocessing     : {df_pre.shape}')
print(f'Pricelist         : {df_price.shape}')
display(df_raw.head())
```

## 2. EDA: Pendapatan

```python
plot_revenue_combo(df_raw)
df_desc = plot_revenue_item(df_raw, df_price)
```

## 3. Transformasi Matriks Biner

```python
df_encoded = create_binary_matrix(transactions)
print(f'Matriks: {df_encoded.shape}')
display(df_encoded.astype(int).head(10))
```

## 4. Eksekusi Apriori

```python
frequent_itemsets = run_apriori(df_encoded, min_support=0.04)
print(f'Frequent itemsets: {len(frequent_itemsets)}')
display(frequent_itemsets)
```

## 5. Association Rules

```python
rules = generate_rules(frequent_itemsets, min_confidence=0.3)
print(f'Rules: {len(rules)}')
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

## 6. Visualisasi

```python
plot_heatmap(rules)
plot_top_rules(rules)
plot_network(rules)
```

## 7. Kesimpulan

```python
print_summary(rules)
```
